## Operadores Lógicos

Em muitos problemas de programação, há necessidade de se fazer **comparações** entre variáveis.

Dica: sempre leia as comparações como uma **pergunta**

Os operadores de conjunção são: **and** e **or**, e eles seguem a seguinte regra:

- **and** só é True se **TODAS** as comparações forem True.

## Condicionais

As condicionais são bem interessantes porque agora podemos tomar decisões com base nos resultados lógicos.

Por exemplo:
Vamos determinar a aprovação de um aluno com base na sua nota final e presença nas aulas.
Vamos considerar que o aluno será aprovado se:
    * Nota final for acima de 7.0
    * Pelo menos 75% de presença

Ou seja, em termos lógicos, o aluno é aprovado **se** tiver `nota` >= 7 e `frequencia` >= 75% (viu onde entra os operadores lógicos?) 😉

### If

O `if` é um operador condicional fundamental em Python. 

A estrutura do if é assim:

```python
if condicao:
  código a ser executado
```

Se a condição fornecida após o `if` for verdadeira, o bloco de código dentro do `if` será executado. Caso contrário, o bloco de código não será executado.

**Exemplo:** Escreva um código que verifique a média do aluno e retorne uma mensagem caso ele seja aprovado.

````python
nota_1 = 80
nota_2 = 60
nota_3 = 70
nota_4 = 40
#Calculando a média simples
media = (nota_1+nota_2+nota_3+nota_4)/4
print('A sua média é:', media)
# Verifica se a media é suficiente para aprovar (maior ou igual a 7)
if media>=7.0:
  # exibe mensagem para aprovar o aluno
  print("O aluno está aprovado por nota")
````

### Else

O ```else``` é usado em conjunto com o if e permite que você execute um bloco de código alternativo quando a condição do if for avaliada como falsa.

Sua estrutura seria:

```python
if condicao:
    codigo_do_if
else:
    codigo_do_else
```

Ou seja, se a condição do if não for verdadeira, o bloco de código dentro do else será executado.

**Exemplo:**

Escrever um programa que define a situação de um aluno conforme as seguintes instruções:

* Média for maior ou igual a 7, o aluno está aprovado.
* **Senão**, aluno está reprovado;

````python
media = 4
if media<7:
  print('Você está reprovado!')
else:
  print('Você está aprovado!')
````

### Elif

O ```elif``` é uma abreviação de "else if". 
<br>Ele é usado para adicionar condições adicionais após um if.
<br>Quando você deseja verificar várias condições em sequência, o elif é útil.

```python
if (condição):
    acao 1
elif (outra_condicao):
    acao 2
elif (outra_condicao):
    acao 3
    ...
else:
    acao final
```

**Obs.:** Ele só é verificado se todas as condições anteriores (incluindo o if e os elifs anteriores) forem falsas.

# Laços While e For

## While

Uma das utilidades de linguagens de programação é a de automatizar tarefas que são repetitivas.

Mas, pra isso ser viável, seria bom se tivéssemos uma estrutura para **repetir comandos**, não é mesmo?

Imagine que eu queira exibir na tela "Olá, mundo!" 5 vezes. Podemos fazer:

````python
#Sem o While
print('Olá, mundo!')
print('Olá, mundo!')
print('Olá, mundo!')
print('Olá, mundo!')
print('Olá, mundo!')
````

Mas, e se eu quiser exibir essa mensagem 1000 vezes? Ou 1 milhão de vezes? Não é ideal escrevermos o mesmo pedaço de código tantas vezes, né?

Para isso, existem os **laços de repetição**, que permitem repetir pedaços de código quantas vezes desejarmos!

O primeiro laço que vamos ver é o **while**. Este laço tem a seguinte estrutura:

```python
while (condicao é True):
    operacao_repetida
```

Ou seja, o que tá no bloco do while é repetido **enquanto a condição for verdadeira**:

````python
#Usando o While
contador = 5
while contador > 0:
  print('Olá, mundo!')
  contador-=1
````

### **ATENÇÃO:**

Você precisa estar atento à **condição**, pois uma condição errada pode levar a **loops infinitos**.

**Veja o exemplo abaixo:**

````python
#loop infinito com while
while 2==2:
  print('Olá, mundo!')
````

Para que loops infinitos não aconteçam, temos que fazer uma **atualização da condição** a cada iteração do laço!

Assim, o que fazemos é **definir a condição do while em termos de uma variável que tenha seu valor atualizado!**

````python
contador = 0

# Quero exibir 5 vezes a mensagem "olá, mundo!"
while contador < 5:
  print('Olá, mundo!')
  # Atualizar o contador
  contador+=1
````

Vamos entender um pouco melhor como a variável contadora se comporta?

Pra isso, basta exibi-la a cada iteração:

````python
#Imprimir o contador
contador = 0

while contador < 5:
  print(contador)
  #print('Olá, mundo!')
  contador+=1
````

Na prática, o que está acontecendo é:

````python
#Inicialmente, a variável contador é 0
contador = 0
print(contador)

#Precisamos atualizar o contador
contador+=1
#Contador vale 1
print(contador)

#Precisamos atualizar o contador
contador+=1
#Contador vale 2
print(contador)

#Precisamos atualizar o contador
contador+=1
#Contador vale 3
print(contador)

#Precisamos atualizar o contador
contador+=1
#Contador vale 4
print(contador)

#Precisamos atualizar o contador
contador+=1
#Contador vale 5
````

EXTRA:
Podemos também atualizar a condição de repetição segundo informado pelo usuário.

Esse uso é bem importante para **garantir que o usuário digitou corretamente o que foi solicitado**.

Por exemplo, vamos pedir pro usuário digitar um número maior que 10. **Enquanto ele não fizer o que queremos**, vamos continuar pedindo pra ele digitar um novo valor:

## For

Agora, veremos um outro laço, o **for**

A estrutura do for é:

```python
for item in iteravel:
    operacao_feita_pra_cada_item
```

O `for` é utilizado para **iterar/percorrer** os elementos de um objeto **iterável** (lista, tupla, texto, etc...)

*Nota*: Objeto iterável possui vários elementos dentro dele, sendo possível percorrer cada elemento numa determinada ordem.

````python
# For em uma lista (objeto que aceita todos os tipos de variáveis)
lista = ["a", "b", "d", True, 1, 3.6]

for elemento in lista:
  print(elemento)
````

O código acima é equivalente a:

````python
#Imprimir sem o for acessando cada elemento individualmente
print(lista[0])
print(lista[1])
print(lista[2])
print(lista[3])
print(lista[4])
print(lista[5])
````

````python
#Podemos fazer para qualquer estrutura iterável
txt = "Olá, Mundo!"

for caractere in txt:
  print(caractere)
````

````python
#Em uma tupla
tupla_1 = ("a",1,3.14,True)

for item in tupla_1:
  print(item)
````

````python
#Em um dicionário
dict = {'Nome':'Romero','Idade':32,'Estado':'ES'}

for key,value in dict.items():
  print(key,value)
````

## Desempacotamento de Tuplas

- O que é desempacotamento de tuplas?
- Como desempacotar valores em variáveis individuais?
- Exemplo prático

````python
# criando tuple
tupla = ('João', 'Silva', 25)

# desempacotando tuple (atribuir uma variavel pra cada)
nome, sobrenome, idade = tupla

# exibir resultado
print(nome)
print(sobrenome)
print(idade)
````

Além disso, podemos usar um único caractere "descartável" (geralmente um sublinhado _) para ignorar elementos da tupla que não são necessários no desempacotamento:

````python
tupla = ('João', 'Silva', 25, 'Brasileiro')

# desempacotando tupla, descartando o item 3o item
nome, sobrenome, _, nacionalidade = tupla

print(nome)
print(sobrenome)
print(nacionalidade)
````

No exemplo acima, estamos desempacotando a tupla e ignorando o terceiro elemento usando o caractere _.

O desempacotamento de tuplas é uma técnica útil quando queremos atribuir os elementos de uma tupla a variáveis individuais de forma rápida e direta. É especialmente útil quando temos funções que retornam múltiplos valores em uma tupla e queremos extrair esses valores para uso posterior no programa.

# Dicionários (hashmap)


A eficiente estrutura de _'hash table'_ de chave-valor do Python é chamada de "dict". Ela é composta sempre por uma sequência de pares de chave-valor (_key-value_).

Num dicionário (o de papel) procuramos alguma palavra e, em seguida, entendemos o significado ou definição dela. O dicionário em python é semelhante, pois quando buscamos a palavra, é como se buscássemos a chave no dicionário. Após encontrá-la, obtemos qual é a definição da palavra. Essa definição é o mesmo que obter o valo da chave (no caso, a definição da palavra.)

Já não precisamos mais procurar alguma coisa pelo seu índice numa lista, o que torna o dicionário um objeto mais estruturado e de rápida localização de itens.



Podemos criar o dicionário dessa forma:

````python
# criar um dict de frutas com key-value fruta: valor da fruta
frutas = {"maçã": 1.50,
          "banana": 0.99,
          "laranja": 1.20}
````

Outra possível forma é utilizara  função `dict` e inserir uma lista de listas de pares de [key,values]:

````python
# criar a lista de pares key,value
lista_keys_values = [["maçã", 1.50],["banana", 0.99],["laranja", 1.20]]

# criar o dict
dict(lista_keys_values)
````

Quando temos os dados de key e value separados em listas, podemos obter uma lista igual a do caso acima utilizando a função `zip` (o que ela faz mesmo?):

````python
# keys e values separados em listas
keys = ["maçã","banana","laranja"]
values = [1.50,0.99,1.20]

# criar uma lista de listas
lista_keys_values = list(zip(keys,values))

# como no caso acima, criar um dict
dict = (lista_keys_values)
````

**Exercício 1:**

Crie dois dicionários abaixo e combine-os em um terceiro dicionário. Se uma chave estiver presente nos dois dicionários originais, some os valores correspondentes.

``` python
dicionario1 = {"a": 1, "b": 2, "c": 3}
dicionario2 = {"b": 4, "d": 5, "e": 6}
```

````python
dicionario1 = {"a": 1, "b": 2, "c": 3}
dicionario2 = {"b": 4, "d": 5, "e": 6}

dicionario3 = dicionario1.copy()
for chave, valor in dicionario2.items():
    if chave in dicionario3:
        dicionario3[chave] += valor
    else:
        dicionario3[chave] = valor

print(dicionario3)
````

**Exercício 2:**

Escreva uma função que receba uma lista como entrada e retorne um dicionário contendo a contagem de cada elemento único na lista.

`lista = [1, 2, 3, 1, 2, 3, 4, 5, 6, 4, 5]`

````python
def contar_elementos(lista):
    contagem = {}
    for elemento in lista:
        if elemento in contagem:
            contagem[elemento] += 1
        else:
            contagem[elemento] = 1
    return contagem

lista = [1, 2, 3, 1, 2, 3, 4, 5, 6, 4, 5]
resultado = contar_elementos(lista)
print(resultado)
````

Escreva uma função que receba um número romano como entrada e retorne o equivalente em número decimal. Use um dicionário para mapear os algarismos romanos para seus valores decimais.

ex:  "MMXXIII" ---> 2023

````python
def romano_para_decimal(numero_romano):
    mapeamento = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}
    decimal = 0
    for i in range(len(numero_romano)):
        if i > 0 and mapeamento[numero_romano[i]] > mapeamento[numero_romano[i-1]]:
            decimal += mapeamento[numero_romano[i]] - 2 * mapeamento[numero_romano[i-1]]
        else:
            decimal += mapeamento[numero_romano[i]]
    return decimal

numero_romano = "MMXXIII"
decimal = romano_para_decimal(numero_romano)
print(decimal)
````

**Exercício 8** *:

Escreva um programa em Python para calcular o valor médio de todos os n-esimos elementos de cada tuple dentro tupla original. Obs: n indica a posição dos elementos dentro das tuplas.

Por exemplo: a primeira média calculada é a media de todos os elementos na 1a posição de cada tupla.

````python
tupla = ((10, 10, 10, 12), (30, 45, 56, 45), (81, 80, 39, 32), (1, 2, 3, 4))

# quantos tuplas dentro da tupla?
nitems = len(tupla)
#quantos elementos dentro de uma sub tupla?
nelem = len(tupla[0])

# criar uma lista com nitems
resultado = [0] * nitems
# percorre cada elemento da tupla
for sub_tupla in tupla:
    for i,num in enumerate(sub_tupla):
        resultado[i]+=num

# aplicar média
for i in range(len(resultado)):
    resultado[i] = resultado[i]/nelem

print(resultado)
````
